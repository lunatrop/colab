## Iterator

### Activation Function

```python
def activation(X):
        return np.where(X >= 0.0, 1, -1) #step activation
        #return(X) #linear activation
```

### Net Input

```python
def net_input(X):
        return np.dot(X, w_[1:]) + w_[0]
```

### Perceptron
$$
\Delta w_i = \eta (y - \hat{y}) x_i
$$

```python
def ppn_train(index):

    xi = X[index]
    yi = y[index]


    z = net_input(xi)
    output = activation(z)

    delta = (yi - output)

    w_[0] += eta * delta
    w_[1:] += eta * xi.T.dot(delta)
    print("w_ : ", w_)
```

### Iteration
TODO: This can associated with n_iter hyperparamameter to run several times <br>
Here we can examine how weights are updated sample by sample.<br>
The learned function $\phi$ is provided as the last line.

```python
index = 0
for index in range(0, len(X)):
    ppn_train(index)
trained_weights = w_
print(trained_weights)
```
